```python
import pandas as pd

data = {
    "Category": ["A", "B", "A", "B", "A", "B", "A", "B"],
    "Store": ["S1", "S1", "S2", "S2", "S1", "S2", "S2", "S1"],
    "Sales": [100, 200, 150, 250, 120, 180, 200, 300],
    "Quantity": [10, 15, 12, 18, 8, 20, 15, 25]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.groupby("Category")["Sales"].sum()

print(result)
```

```python
result = df.groupby(["Category", "Store"])["Sales"].agg(
    ["sum", "mean", "max", "min"]
)

print(result)
```

```python
grouped = df.groupby("Store")["Quantity"].sum()

print(grouped)

best_store = grouped.idxmax()

print("Best Store:", best_store)
```

```python
result = df.groupby("Category").agg({
    "Sales": ["sum", "mean", "max", "min"],
    "Quantity": ["sum", "mean"]
})

print(result)
```
